## 🧩Paso 4: Summary statistics de uso por usuario


### 4.1 Agrupación por comportamiento de uso



```python
# Columnas auxiliares
usage["is_text"] = (usage["type"] == "text").astype(int) #conocer el total de mensajes
usage["is_call"] = (usage["type"] == "call").astype(int) #conocer el total de llamadas


# Agrupar información por usuario
usage_agg = usage.groupby('user_id').agg({'is_text': 'sum', 'is_call':'sum', 'duration':'sum'}).reset_index()

# observar resultado
usage_agg.head()
```

```python
# Renombrar columnas
usage_agg.columns = ['user_id','cant_mensajes', 'cant_llamadas', 'cant_minutos_llamada']
# observar resultado
usage_agg.head()
```

```python
# Combinar la tabla agregada con el dataset de usuarios
user_profile = usage_agg.merge(users, on= 'user_id', how='left')
user_profile.head()
```

---
**Hallazgos de agrupación por comportamiento de uso**

Se verificó que el merge se realizó correctamente, conservando las transformaciones previas (como los NaN en city) y agregando las métricas de uso (cant_mensajes, cant_llamadas, cant_minutos_llamada) a la tabla de usuarios.

```python
# Calcular límites con el método IQR
columnas_limites = ['cant_mensajes', 'cant_llamadas','cant_minutos_llamada']

for col in columnas_limites:
    Q1 = user_profile[col].quantile(0.25)
    Q3 = user_profile[col].quantile(0.75)

    IQR = Q3 - Q1

    limite_superior = Q3 +1.5 * IQR

    print(f'\n{col}:')
    print(f'Q1: {Q1}')
    print(f'Q3: {Q3}')
    print(f'IQR: {IQR}')
    print(f'Límite superior: {limite_superior}')
    print(f'valor máximo en datos: {user_profile[col].max()}')


```

Para detectar outliers se utilizó el método IQR (rango intercuartílico), que es una forma robusta de identificar valores atípicos sin depender de la media, ya que esta puede verse afectada por los mismos valores extremos. El IQR se calcula como la diferencia entre el percentil 75 (Q3) y el percentil 25 (Q1), y representa el rango donde se concentra el 50% central de los datos. Cualquier valor que supere el límite superior (Q3 + 1.5 × IQR) o esté por debajo del límite inferior (Q1 − 1.5 × IQR) se considera un outlier, ya que se aleja significativamente del comportamiento típico de los datos.

## 🧩Paso 6: Segmentación de Clientes

### 6.1 Segmentación de Clientes Por Uso

  El objetivo de esta sección es clasificar a los usuarios según su nivel de actividad, basándonos en la cantidad de llamadas y mensajes registrados, para identificar patrones de uso que permitan diseñar estrategias comerciales diferenciadas. Se definieron tres grupos:

  - usuarios con menos de 5 llamadas y menos de 5 mensajes se consideran de `'bajo uso'`
  - usuarios con menos de 10 llamadas y menos de 10 mensajes se consideran de `'uso medio'`
  - el resto se clasifica como de `'alto uso'`

```python
# Crear columna grupo_uso
def grupo_uso(row):
    if (row['cant_llamadas'] < 5) & (row['cant_mensajes'] <5):
        return 'Bajo uso'
    elif (row['cant_llamadas']<10) & (row['cant_mensajes']<10):
        return 'Uso medio'
    else:
        return 'Alto uso'
user_profile['grupo_uso'] = user_profile.apply(grupo_uso, axis=1)
```

### 6.2 Segmentación de Clientes Por Edad

El objetivo de esta sección es clasificar a los usuarios según su rango de edad, para identificar patrones generacionales que permitan diseñar estrategias comerciales diferenciadas. Se definieron tres grupos:

- usuarios menores de 30 años se consideran `'jóvenes'`
- usuarios menores de 60 años se consideran `'adultos'`
- el resto se clasifica como `'adulto mayor'`

```python
# Crear columna grupo_edad
def grupo_edad(row):
    if (row['age']<30):
        return 'Joven'
    elif (row['age']<60):
        return 'Adulto'
    else:
        return 'Adulto Mayor'

user_profile ['grupo_edad'] = user_profile.apply(grupo_edad, axis=1)
```
